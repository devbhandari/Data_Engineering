```python
# --using .venv_dbx env
from pyspark.sql.functions import *

df = (
    spark.readStream
    .format("csv")
    .option("header", "true")
    .schema("""
        tpep_pickup_datetime STRING,
        tpep_dropoff_datetime STRING,
        trip_distance DOUBLE,
        fare_amount DOUBLE,
        pickup_zip INT,
        dropoff_zip INT
    """)
    .load("/databricks-datasets/nyctaxi/")
)

query = (
    df.writeStream
    .format("parquet")
    .outputMode("append")
    .option("checkpointLocation", "/tmp/checkpoints/nyc_taxi")
    .start("/tmp/output/nyc_taxi")
)

query.awaitTermination(30)
```
